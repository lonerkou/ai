```python
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense

import numpy as np

# ==========================================
# 1. 準備模擬資料集 (Synthetic Data)
# ==========================================

# 隨機產生 1000 筆資料，每筆資料包含 10 個浮點數特徵 (features)
data = np.random.random((1000, 10))

# 隨機產生 1000 筆對應的標籤 (labels)，數值上限為 2 (即 0 或 1，代表二元分類的類別)
labels = np.random.randint(2, size=(1000, 1))

# 印出資料與標籤的形狀 (Shape) 以及第一筆資料內容
print("--- 資料與標籤資訊 ---")
print("Data shape:", np.shape(data))
print("Data[0]:", data[0])
print("Labels shape:", np.shape(labels))
print("Labels[0]:", labels[0])
print("-" * 30)

# ==========================================
# 2. 建立深度學習模型架構 (Sequential 序列模型)
# ==========================================
model = Sequential()

# 加入第一層（隱藏層）：32個神經元，激勵函數用 'relu'，輸入維度為 10
model.add(Dense(32, activation='relu', input_dim=10))

# 加入第二層（輸出層）：1個神經元，激勵函數用 'sigmoid'（輸出 0~1 的機率值，適合二元分類）
model.add(Dense(1, activation='sigmoid'))

# ==========================================
# 3. 編譯模型 (Compile)
# ==========================================
# - optimizer: 最佳化演算法 ('rmsprop')
# - loss: 損失函數，使用二元交叉熵 ('binary_crossentropy')
# - metrics: 評估模型的指標，這裡追蹤正確率 ('accuracy')
model.compile(optimizer='rmsprop', loss='binary_crossentropy', metrics=['accuracy'])

# ==========================================
# 4. 訓練模型 (Fit)
# ==========================================
# 將資料代入模型訓練：epochs=10 (訓練 10 個完整週期), batch_size=32 (每批次 32 筆資料)
model.fit(data, labels, epochs=10, batch_size=32)

# ==========================================
# 5. 評估模型 (Evaluate)
# ==========================================
# 評估模型在訓練資料上的表現，回傳 [loss, accuracy]
score = model.evaluate(data, labels)

# 印出模型的最終正確率 (accuracy)
print("模型最終正確率 (Accuracy):", score[1])
```
